```python
def apply_mask_conditions(mask_array, mask_conditions):
    """
    Apply mask conditions to create a boolean mask.
    
    Parameters
    ----------
    mask_array : numpy.ndarray
        Array containing integer mask values
    mask_conditions : dict or list
        If dict: mapping of condition names to values, e.g., {'shifted':1, 'array':2}
        If list: list of values to include, e.g., [1, 2]
        
    Returns
    -------
    numpy.ndarray
        Boolean mask where True indicates points to include
    """
    if mask_conditions is None:
        # No masking - include all points
        return np.ones_like(mask_array, dtype=bool)
    
    if isinstance(mask_conditions, dict):
        # Extract values from dictionary
        valid_values = list(mask_conditions.values())
    elif isinstance(mask_conditions, (list, tuple)):
        valid_values = list(mask_conditions)
    else:
        raise ValueError("mask_conditions must be a dict, list, or tuple")
    
    # Create boolean mask for valid points
    bool_mask = np.isin(mask_array, valid_values)
    
    return bool_mask

bool_mask = apply_mask_conditions(ds.mask.values,[1,2])
print(bool_mask.astype(float).min())
```

# Get second order structure function

General Form
The conditional structure function is defined as:
$Dij(αβ)(x,r)=1Pαβ⟨[ui(x+r,t)−ui(x,t)][uj(x+r,t)−uj(x,t)]Iα(x,t)Iβ(x+r,t)⟩D_{ij}^{(\alpha\beta)}(\mathbf{x}, \mathbf{r}) = \frac{1}{P_{\alpha\beta}} \langle [u_i(\mathbf{x}+\mathbf{r},t) - u_i(\mathbf{x},t)][u_j(\mathbf{x}+\mathbf{r},t) - u_j(\mathbf{x},t)] I_\alpha(\mathbf{x},t) I_\beta(\mathbf{x}+\mathbf{r},t) \rangleDij(αβ)​(x,r)=Pαβ​1​⟨[ui​(x+r,t)−ui​(x,t)][uj​(x+r,t)−uj​(x,t)]Iα​(x,t)Iβ​(x+r,t)⟩$
Where:

Iα(x,t)I_\alpha(\mathbf{x},t)
Iα​(x,t) is an indicator function that equals 1 if condition α\alpha
α is satisfied at position x\mathbf{x}
x and time tt
t, and 0 otherwise

Iβ(x+r,t)I_\beta(\mathbf{x}+\mathbf{r},t)
Iβ​(x+r,t) is an indicator function that equals 1 if condition β\beta
β is satisfied at position x+r\mathbf{x}+\mathbf{r}
x+r and time tt
t, and 0 otherwise

Pαβ=⟨Iα(x,t)Iβ(x+r,t)⟩P_{\alpha\beta} = \langle I_\alpha(\mathbf{x},t) I_\beta(\mathbf{x}+\mathbf{r},t) \rangle
Pαβ​=⟨Iα​(x,t)Iβ​(x+r,t)⟩ is the probability that conditions α\alpha
α and β\beta
β are satisfied at positions x\mathbf{x}
x and x+r\mathbf{x}+\mathbf{r}
x+r respectively

# Distribution of lowest bin (smallest separation distance) uncertainties

```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

# Configuration
nb_values = np.array([10, 20, 30, 50, 75, 100, 150, 200, 300, 400, 500])
n_realizations = 30  # Independent runs at each nb to build distribution

# Focus on smallest scale
bin_r = np.logspace(np.log10(1.0e4), np.log10(1.0e5), 10)
bins = {'r': bin_r}

# Store results: for each nb, collect multiple uncertainty estimates
uncertainty_distribution = np.zeros((len(nb_values), n_realizations))

for i, nb in enumerate(nb_values):
    print(f"\n=== Testing nb = {nb} ===")
    
    for j in range(n_realizations):
        print(f"  Realization {j+1}/{n_realizations}")
        
        iso_sf_bootstrap = psf.get_isotropic_sf_2d(
            ds=dsh,
            variables_names=["u","v"],
            order=2,
            bins=bins,
            fun='default_vel',
            bootsize={'y':159,'x':120},
            initial_nbootstrap=nb,
            max_nbootstrap=nb,
            step_nbootstrap=0,
            convergence_eps=0.003,
            n_jobs=-1,
            backend='loky'
        )
        
        uncertainty_distribution[i, j] = iso_sf_bootstrap['std'][0]
        gc.collect()
    
    # Statistics at this nb
    mean_unc = np.mean(uncertainty_distribution[i, :])
    std_unc = np.std(uncertainty_distribution[i, :], ddof=1)
    print(f"  Mean uncertainty: {mean_unc:.4e}")
    print(f"  Std of uncertainty: {std_unc:.4e}")

# Analyze convergence using CLT
mean_uncertainty = np.mean(uncertainty_distribution, axis=1)
std_uncertainty = np.std(uncertainty_distribution, axis=1, ddof=1)
sem_uncertainty = std_uncertainty / np.sqrt(n_realizations)

# CLT predicts: std(uncertainty) ~ A/√nb + B
def clt_model(nb, A, B):
    return A / np.sqrt(nb) + B

# Fit the model
popt, pcov = curve_fit(clt_model, nb_values, std_uncertainty, 
                       p0=[100, 0.001])
A_fit, B_fit = popt

# Find optimal nb (e.g., where improvement < 5%)
threshold = 0.05
std_asymptotic = B_fit
for i in range(len(nb_values)-1):
    improvement = (std_uncertainty[i] - std_uncertainty[i+1]) / std_uncertainty[i]
    if improvement < threshold:
        optimal_nb = nb_values[i]
        break
else:
    optimal_nb = nb_values[-1]

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Convergence of std(uncertainty)
axes[0].errorbar(nb_values, std_uncertainty, yerr=sem_uncertainty, 
                 fmt='o', capsize=5, label='Empirical', markersize=8)
nb_fine = np.linspace(nb_values[0], nb_values[-1], 200)
axes[0].plot(nb_fine, clt_model(nb_fine, A_fit, B_fit), 
             'r--', lw=2, label=f'CLT fit: {A_fit:.2f}/√nb + {B_fit:.4f}')
axes[0].axvline(optimal_nb, color='green', ls=':', lw=2, 
                label=f'Optimal nb ≈ {optimal_nb}')
axes[0].set_xlabel('Number of Bootstraps (nb)', fontsize=12)
axes[0].set_ylabel('Std(Uncertainty Estimate)', fontsize=12)
axes[0].set_title('CLT Convergence Analysis', fontsize=13)
axes[0].legend()
axes[0].grid(alpha=0.3)
axes[0].set_xscale('log')
axes[0].set_yscale('log')

# Plot 2: Mean uncertainty (should be stable)
axes[1].errorbar(nb_values, mean_uncertainty, yerr=std_uncertainty,
                 fmt='s', capsize=5, markersize=8, color='blue')
axes[1].axhline(np.mean(mean_uncertainty), color='red', ls='--', 
                label='Overall mean')
axes[1].axvline(optimal_nb, color='green', ls=':', lw=2,
                label=f'Optimal nb ≈ {optimal_nb}')
axes[1].set_xlabel('Number of Bootstraps (nb)', fontsize=12)
axes[1].set_ylabel('Mean Uncertainty Estimate', fontsize=12)
axes[1].set_title('Stability of Mean Estimate', fontsize=13)
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('clt_bootstrap_convergence.png', dpi=300, bbox_inches='tight')

print(f"\n=== CLT ANALYSIS RESULTS ===")
print(f"Fitted model: std(unc) = {A_fit:.2f}/√nb + {B_fit:.4f}")
print(f"Asymptotic std: {B_fit:.4e}")
print(f"Optimal nb (< {threshold*100}% improvement): {optimal_nb}")
print(f"Std at optimal nb: {std_uncertainty[nb_values==optimal_nb][0]:.4e}")
```
