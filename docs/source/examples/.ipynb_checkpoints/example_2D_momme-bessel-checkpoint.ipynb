# 2D Structure Function



```python
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm, SymLogNorm
import matplotlib.ticker as ticker
import pyturbo_sf as psf
import pyproj
import numpy as np
linewidth = 2
fontsize = 12
plt.rcParams['xtick.labelsize'] = fontsize
plt.rcParams['ytick.labelsize'] = fontsize
plt.rcParams['xtick.major.width'] = 2
plt.rcParams['xtick.minor.width'] = 2
plt.rcParams['ytick.major.width'] = 2
plt.rcParams['ytick.minor.width'] = 2
plt.rcParams['xtick.major.size'] = 10
plt.rcParams['xtick.minor.size'] = 5
plt.rcParams['ytick.major.size'] = 10
plt.rcParams['ytick.minor.size'] = 5
plt.rcParams['savefig.dpi'] = 150
plt.rc('font', family='serif')
import gc
```

# Load the Dataset

```python
path_data = 'example_data/'
dsh = xr.open_dataset(path_data+'/'+'Gi_example_output.nc')
```

```python
dsh
```

# Make A quick Plot

```python
fig,ax = plt.subplots(figsize=(6,6))
im = ax.pcolormesh(dsh.x/1.0e3,dsh.y/1.0e3,0.5*(dsh.u**2 + dsh.v**2),cmap='jet',shading='gouraud')

pos1cb = ax.get_position()
cbar_ax = fig.add_axes([pos1cb.x0+pos1cb.width/4., pos1cb.y0+0.04+pos1cb.height , pos1cb.width/2. , 0.009])
cbar = fig.colorbar(im, cax=cbar_ax,orientation='horizontal',\
                            extend='both',ticks=np.arange(0,1,0.2))
cbar.set_label(r'$ {\rm KE \,  [} {\rm m^{2}~} {\rm s^{-2}]}$',fontsize= fontsize + 2 ,labelpad=10)
cbar.ax.tick_params(labelsize=fontsize,direction='in',width=linewidth,size=4)
cbar.ax.xaxis.set_label_position('top')
cbar.ax.xaxis.set_ticks_position('both')
cbar.outline.set_linewidth( linewidth )
for axis in ['top','bottom','left','right']:
    ax.spines[axis].set_linewidth( linewidth )
    ax.spines[axis].set_color('k')
ax.tick_params(direction='in',bottom='on',top='on',left='on',right='on')
ax.set_xlabel('X [km]')
ax.set_ylabel('Y [km]')
```

# Calculate ISOTROPIC Advective sf 

## No Bootstrapping performed ie., bootsize = data size 

```python
%%time

nbins = np.round(np.log2(240)+1) # 240 is the smallest dim here -- sturge rule

bin_r = np.logspace(np.log10(1.0e4), np.log10(5.0e5), int(nbins+1))
bins = {
    'r': bin_r, # in m
}
iso_sf_adv = psf.get_isotropic_sf_2d(
    ds=dsh,
    variables_names=["u","v","adv_u","adv_v"],
    order=1,
    bins=bins,
    fun='advective',
    bootsize={'y':319,'x':240},
)

bin_r = np.logspace(np.log10(1.0e4), np.log10(5.0e5), 15)
bins = {
    'r': bin_r, # in m
}
iso_sf_adv_15 = psf.get_isotropic_sf_2d(
    ds=dsh,
    variables_names=["u","v","adv_u","adv_v"],
    order=1,
    bins=bins,
    fun='advective',
    bootsize={'y':319,'x':240},
)

bin_r = np.logspace(np.log10(1.0e4), np.log10(5.0e5), 20)
bins = {
    'r': bin_r, # in m
}
iso_sf_adv_20 = psf.get_isotropic_sf_2d(
    ds=dsh,
    variables_names=["u","v","adv_u","adv_v"],
    order=1,
    bins=bins,
    fun='advective',
    bootsize={'y':319,'x':240},
)
```

```python
# Create separate figures for better spacing and clarity
plt.figure(figsize=(10, 6))
# Structure Function with Confidence Interval
# Add longitudinal structure function with error bars
plt.plot(iso_sf_adv.r, iso_sf_adv.sf, 'b-', linewidth=2.5, label='iso_sf_adv (nbins=10)')
plt.title('iso_sf_adv')
plt.legend()
plt.xscale('log')


# Create separate figures for better spacing and clarity
plt.figure(figsize=(10, 6))
# Structure Function with Confidence Interval
# Add longitudinal structure function with error bars
plt.plot(iso_sf_adv_15.r, iso_sf_adv_15.sf, 'b-', linewidth=2.5, label='iso_sf_adv (nbins=15)')
plt.title('iso_sf_adv')
plt.legend()
plt.xscale('log')

# Structure Function with Confidence Interval
plt.figure(figsize=(10, 6))

# Add longitudinal structure function with error bars
plt.plot(iso_sf_adv_20.r, iso_sf_adv_20.sf, 'b-', linewidth=2.5, label='iso_sf_adv (nbins=20)')
plt.title('iso_sf_adv')
plt.legend()
plt.xscale('log')
```

# Calculate ISOTROPIC Pwork sf using fun="pressure_work"

## No Bootstrapping performed ie., bootsize = data size 

```python
%%time

nbins = np.round(np.log2(240)+1) # 240 is the smallest dim here -- sturge rule

bin_r = np.logspace(np.log10(1.0e4), np.log10(5.0e5), int(nbins+1))
bins = {
    'r': bin_r, # in m
}
iso_sf_adv = psf.get_isotropic_sf_2d(
    ds=dsh,
    variables_names=["p","u","v"],
    order=1,
    bins=bins,
    fun='pressure_work',
    bootsize={'y':319,'x':240},
)

bin_r = np.logspace(np.log10(1.0e4), np.log10(5.0e5), 15)
bins = {
    'r': bin_r, # in m
}
iso_sf_adv_15 = psf.get_isotropic_sf_2d(
    ds=dsh,
    variables_names=["p","u","v"],
    order=1,
    bins=bins,
    fun='pressure_work',
    bootsize={'y':319,'x':240},
)

bin_r = np.logspace(np.log10(1.0e4), np.log10(5.0e5), 20)
bins = {
    'r': bin_r, # in m
}
iso_sf_adv_20 = psf.get_isotropic_sf_2d(
    ds=dsh,
    variables_names=["p","u","v"],
    order=1,
    bins=bins,
    fun='pressure_work',
    bootsize={'y':319,'x':240},
)
```

```python
# Create separate figures for better spacing and clarity
plt.figure(figsize=(10, 6))
# Structure Function with Confidence Interval
# Add longitudinal structure function with error bars
plt.plot(iso_sf_adv.r, iso_sf_adv.sf, 'b-', linewidth=2.5, label='iso_sf_pwork (nbins=10)')
plt.title('iso_sf_Pwork')
plt.xscale('log')
plt.legend()


# Create separate figures for better spacing and clarity
plt.figure(figsize=(10, 6))
# Structure Function with Confidence Interval
# Add longitudinal structure function with error bars
plt.plot(iso_sf_adv_15.r, iso_sf_adv_15.sf, 'b-', linewidth=2.5, label='iso_sf_pwork (nbins=15)')
plt.title('iso_sf_Pwork')
plt.xscale('log')

plt.legend()

# Structure Function with Confidence Interval
plt.figure(figsize=(10, 6))

# Add longitudinal structure function with error bars
plt.plot(iso_sf_adv_20.r, iso_sf_adv_20.sf, 'b-', linewidth=2.5, label='iso_sf_pwork (nbins=20)')
plt.title('iso_sf_Pwork')
plt.xscale('log')
plt.legend()
```
