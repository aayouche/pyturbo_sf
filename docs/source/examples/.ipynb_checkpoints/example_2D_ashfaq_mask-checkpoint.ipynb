```python
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm, SymLogNorm
import matplotlib.ticker as ticker
import pyturbo_sf as psf
import pyproj
import numpy as np
linewidth = 2
fontsize = 12
plt.rcParams['xtick.labelsize'] = fontsize
plt.rcParams['ytick.labelsize'] = fontsize
plt.rcParams['xtick.major.width'] = 2
plt.rcParams['xtick.minor.width'] = 2
plt.rcParams['ytick.major.width'] = 2
plt.rcParams['ytick.minor.width'] = 2
plt.rcParams['xtick.major.size'] = 10
plt.rcParams['xtick.minor.size'] = 5
plt.rcParams['ytick.major.size'] = 10
plt.rcParams['ytick.minor.size'] = 5
plt.rcParams['savefig.dpi'] = 150
plt.rc('font', family='serif')
import gc
```

# Load Dataset

```python
idt = 10

dsu = xr.open_dataset('/home/aayouche/Downloads/FRAM_u.nc').isel(time=idt)
dsv = xr.open_dataset('/home/aayouche/Downloads/FRAM_v.nc').isel(time=idt)
dsg = xr.open_dataset('/home/aayouche/Downloads/FRAM_grid.nc')

# Convert Latitude and Loongitude to x and y

lon = dsg.lon
lat = dsg.lat

lon_center = np.nanmin(lon)
utm_zone = int(1 + (lon_center + 180.0) / 6.0)

# Create the projection
proj = pyproj.Proj(proj='utm', zone=utm_zone, ellps='WGS84', datum='WGS84')

x, y = proj(lon.values, lat.values)

y_dim,x_dim = x.shape
# Create X and Y DataArrays with 2D structure
X = xr.DataArray(
    x,
    dims=['y', 'x'],
    coords={
        'y': np.arange(y_dim),
        'x': np.arange(x_dim)
    }
)

Y = xr.DataArray(
    y,
    dims=['y', 'x'],
    coords={
        'y': np.arange(y_dim),
        'x': np.arange(x_dim)
    }
)

# rename dimensions for u and v

u = dsu.U.rename({'j':'y','i':'x'})
v = dsv.V.rename({'j':'y','i':'x'})

```

# Format Dataset for Pyturbo

```python
ds = xr.Dataset(
    data_vars={
        'u': u,
        'v': v,
    },
    coords={
        'x': X,
        'y': Y,
    }
)
```

```python
ds
```

# Quick Plot

```python
plt.pcolormesh((x-x.min())/1e3,(y-y.min())/1e3,ds.u)
plt.colorbar()
```

# Compute Structure Function Second Order

```python
nbins = 11  # Increase for better resolution of scaling behavior
bin_r = np.logspace(np.log10(1.0e3), np.log10(1.0e5), int(nbins+1))
bins = {
    'r': bin_r,  # in m
}


iso_sf_bootstrap = psf.get_isotropic_sf_2d(
    ds=ds,
    variables_names=["u","v"],
    order=2,
    bins=bins,
    fun='default_vel',
    bootsize={'y':88,'x':62},
    initial_nbootstrap=4000,
    max_nbootstrap=4000,
    step_nbootstrap=0,
    convergence_eps=0.05,
    n_jobs=-1,
    backend='loky'
)
```

# Plot Second Order Structure Function

```python
import numpy as np
from scipy import stats

plt.figure(figsize=(8, 6))
plt.loglog(iso_sf_bootstrap.r/1e3, iso_sf_bootstrap.sf, '-o', label='bootstrap')
plt.fill_between(iso_sf_bootstrap.r/1.0e3, iso_sf_bootstrap.ci_lower, iso_sf_bootstrap.ci_upper, alpha=0.3)

# Compute the actual slope from the data (excluding first and last points)
log_r = np.log10(iso_sf_bootstrap.r[1:-1])
log_sf = np.log10(iso_sf_bootstrap.sf[1:-1])
slope, intercept, r_value, p_value, std_err = stats.linregress(log_r, log_sf)

# Plot the fitted line
r_fit = iso_sf_bootstrap.r/1e3
fitted_line = 10**intercept * (r_fit*1e3)**slope
plt.loglog(r_fit, fitted_line, 'g-', alpha=0.7, linewidth=2, 
           label=f'Fitted slope: {slope:.2f} (R²={r_value**2:.3f})')

# Add Kolmogorov 2/3 slope reference line
r_ref = iso_sf_bootstrap.r/1e3  # in km
mid_idx = len(r_ref)//2
amplitude = iso_sf_bootstrap.sf[mid_idx] / (r_ref[mid_idx]**(2/3))
kolmogorov_line = amplitude * r_ref**(2/3)
plt.loglog(r_ref, kolmogorov_line, 'k--', alpha=0.7, label=r'$r^{2/3}$ (Kolmogorov)')

plt.xlabel('Separation distance [km]')
plt.ylabel(r'$\delta u^2 + \delta v^2~[\rm m^{2}~s^{-2}]$')
plt.legend()
plt.grid(True, which="both", alpha=0.3)
plt.title(f'Structure Function - Computed slope: {slope:.3f} vs Kolmogorov: 0.667')
```
